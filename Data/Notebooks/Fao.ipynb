```python
import rasterio
from rasterio.windows import from_bounds
import numpy as np

input_file = "/content/WHE_HILM.tif"
output_file = "/content/WHE_HILM_Zankaloun.tif"

# Approximate bounding box of El-Zankaloun
west, south, east, north = 31.35, 30.44, 31.55, 30.64

with rasterio.open(input_file) as src:
    window = from_bounds(
        west, south, east, north,
        src.transform
    )

    window = window.round_offsets().round_lengths()

    data = src.read(1, window=window)
    transform = src.window_transform(window)

    profile = src.profile.copy()
    profile.update(
        height=data.shape[0],
        width=data.shape[1],
        transform=transform
    )

    with rasterio.open(output_file, "w", **profile) as dst:
        dst.write(data, 1)

    valid = data[data != 0]

    print("Output shape:", data.shape)
    print("Valid pixels:", len(valid))
    print("Unique values:", np.unique(valid))
    print("Min:", valid.min() if len(valid) else None)
    print("Max:", valid.max() if len(valid) else None)

print("Saved:", output_file)
```

```python

from google.cloud import storage
import rasterio
from rasterio.windows import from_bounds
import numpy as np
import pandas as pd
import os

client = storage.Client.create_anonymous_client()
bucket = client.bucket("fao-gismgr-gaez-v5-data")

crops = ["WHE", "MZE", "RCD", "RCW", "PEA", "PHB", "COW"]
levels = ["HILM", "LILM", "HRLM", "LRLM"]

prefix = "DATA/GAEZ-V5/MAPSET/RES05-SIX30AS/"

west, south, east, north = 31.35, 30.44, 31.55, 30.64

results = []

for crop in crops:
    for level in levels:

        filename = (
            f"GAEZ-V5.RES05-SIX30AS.HP0120."
            f"AGERA5.HIST.{crop}.{level}.tif"
        )

        path = f"/content/{crop}_{level}.tif"

        bucket.blob(prefix + filename).download_to_filename(path)

        with rasterio.open(path) as src:
            window = from_bounds(
                west, south, east, north, src.transform
            )
            window = window.round_offsets().round_lengths()

            data = src.read(1, window=window)

        valid = data[data != 0]
        values, counts = np.unique(valid, return_counts=True)

        for value, count in zip(values, counts):
            results.append({
                "Crop_Code": crop,
                "Scenario": level,
                "Suitability_Class": int(value),
                "Pixels": int(count),
                "Percentage": round(count / len(valid) * 100, 2)
            })

        print(f"Finished: {crop} - {level}")

        if os.path.exists(path):
            os.remove(path)

df = pd.DataFrame(results)
df.to_csv("/content/GAEZ_Crop_Suitability.csv", index=False)

print("\nAll files processed!")
display(df)
```

```python

summary = (
    df.assign(
        Land_Pixels=np.where(
            df["Suitability_Class"] != 9,
            df["Pixels"],
            0
        )
    )
    .groupby(["Crop_Code", "Scenario"])
    .agg(
        Total_Land_Pixels=("Land_Pixels", "sum")
    )
    .reset_index()
)

df_corrected = df.merge(
    summary,
    on=["Crop_Code", "Scenario"]
)

df_corrected["Land_Percentage"] = np.where(
    df_corrected["Suitability_Class"] != 9,
    df_corrected["Pixels"] / df_corrected["Total_Land_Pixels"] * 100,
    0
)

df_corrected.to_csv(
    "/content/GAEZ_Crop_Suitability_All_Scenarios_Corrected.csv",
    index=False
)

print("Correction completed!")
display(df_corrected)
```

```python

df_corrected["Suitable_Pixels"] = np.where(
    df_corrected["Suitability_Class"].between(1, 7),
    df_corrected["Pixels"],
    0
)

df_corrected["Not_Suitable_Pixels"] = np.where(
    df_corrected["Suitability_Class"] == 8,
    df_corrected["Pixels"],
    0
)

df_corrected["Water_Pixels"] = np.where(
    df_corrected["Suitability_Class"] == 9,
    df_corrected["Pixels"],
    0
)

final_summary = (
    df_corrected.groupby(["Crop_Code", "Scenario"])
    .agg(
        Total_Land_Pixels=("Total_Land_Pixels", "first"),
        Suitable_Pixels=("Suitable_Pixels", "sum"),
        Not_Suitable_Pixels=("Not_Suitable_Pixels", "sum"),
        Water_Pixels=("Water_Pixels", "sum")
    )
    .reset_index()
)

final_summary["Suitable_Percentage"] = (
    final_summary["Suitable_Pixels"] /
    final_summary["Total_Land_Pixels"] * 100
).round(2)

final_summary["Not_Suitable_Percentage"] = (
    final_summary["Not_Suitable_Pixels"] /
    final_summary["Total_Land_Pixels"] * 100
).round(2)

final_summary.to_csv(
    "/content/GAEZ_Final_Summary.csv",
    index=False
)

display(final_summary)
```

```python
class_check = (
    df_corrected
    .groupby(["Crop_Code", "Scenario", "Suitability_Class"])
    ["Pixels"]
    .sum()
    .reset_index()
)

display(
    class_check.sort_values(
        ["Crop_Code", "Scenario", "Suitability_Class"]
    )
)
```
