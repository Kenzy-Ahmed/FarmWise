```python
import pandas as pd

df = pd.read_csv(
    r"G:\FarmWise\Data\Merged_Data\FarmWise_Unified_Dataset.csv"
)

print(df.shape)
print(df.columns.tolist())
```

```python
# Select environmental features
features = [
    'power_t2m_mean_c',
    'power_t2m_max_c',
    'power_t2m_min_c',
    'power_precip_mm_day',
    'gpm_precip_mm_day',
    'power_rh2m_pct',
    'smap_soil_moisture_am_m3m3',
    'soil_ph',
    'soil_soc_g_kg',
    'soil_clay_pct',
    'soil_sand_pct',
    'soil_silt_pct',
    'soil_cec_cmolc_kg',
    'soil_nitrogen_g_kg',
    'ndvi_mean'
]

# Check missing values
print("Missing Values:")
print(df[features].isnull().sum())

# Statistical summary
display(df[features].describe().T)
```

```python
# Check dataset time coverage
df['date'] = pd.to_datetime(df['date'])

print("Start date:", df['date'].min())
print("End date:", df['date'].max())
print("Number of years:", df['year'].nunique())

# Check whether soil features are constant
soil_features = [
    'soil_ph',
    'soil_soc_g_kg',
    'soil_clay_pct',
    'soil_sand_pct',
    'soil_silt_pct',
    'soil_cec_cmolc_kg',
    'soil_nitrogen_g_kg'
]

print("\nUnique values in soil features:")
print(df[soil_features].nunique())
```

```python
annual_climate = df.groupby('year').agg(
    mean_temp=('power_t2m_mean_c', 'mean'),
    avg_max_temp=('power_t2m_max_c', 'mean'),
    avg_min_temp=('power_t2m_min_c', 'mean'),
    annual_rainfall=('power_precip_mm_day', 'sum'),
    avg_humidity=('power_rh2m_pct', 'mean')
).reset_index()

display(annual_climate.round(2))
```

```python
# Monthly climate analysis

monthly_rain = (
    df.groupby(['year', 'month'])['power_precip_mm_day']
    .sum()
    .groupby('month')
    .mean()
)

monthly_temp = df.groupby('month').agg(
    mean_temp=('power_t2m_mean_c', 'mean'),
    max_temp=('power_t2m_max_c', 'mean'),
    min_temp=('power_t2m_min_c', 'mean')
)

monthly_climate = monthly_temp.copy()
monthly_climate['avg_monthly_rainfall'] = monthly_rain

display(monthly_climate.round(2))
```

```python
# Load crop requirements
crops = pd.read_csv(
    r"G:\FarmWise\Data\Crop_Requirements\FarmWise_Crop_Requirements_Complete.csv"
)

# Match monthly average temperature with crop requirements
results = []

for _, crop in crops.iterrows():

    suitable_months = monthly_climate[
        (monthly_climate['mean_temp'] >= crop['temp_opt_min']) &
        (monthly_climate['mean_temp'] <= crop['temp_opt_max'])
    ].index.tolist()

    results.append({
        'crop_name': crop['crop_name'],
        'suitable_temp_months': suitable_months,
        'number_of_months': len(suitable_months)
    })

temperature_matching = pd.DataFrame(results)

display(temperature_matching)
```

```python
temperature_matching['temp_score'] = (
    temperature_matching['number_of_months'] / 12 * 100
).round(2)

temperature_matching['temp_status'] = temperature_matching[
    'temp_score'
].apply(
    lambda x: 'High' if x >= 60
    else 'Moderate' if x >= 40
    else 'Low'
)

display(
    temperature_matching[
        ['crop_name', 'number_of_months', 'temp_score', 'temp_status']
    ]
)
```

```python
rainfall_results = []

for _, crop in crops.iterrows():

    min_rain = crop['rainfall_opt_min_annual_mm']
    max_rain = crop['rainfall_opt_max_annual_mm']

    suitable_years = annual_climate[
        (annual_climate['annual_rainfall'] >= min_rain) &
        (annual_climate['annual_rainfall'] <= max_rain)
    ]['year'].tolist()

    rainfall_results.append({
        'crop_name': crop['crop_name'],
        'suitable_years': suitable_years,
        'number_of_years': len(suitable_years)
    })

rainfall_matching = pd.DataFrame(rainfall_results)

rainfall_matching['rainfall_score'] = (
    rainfall_matching['number_of_years'] / len(annual_climate) * 100
).round(2)

display(rainfall_matching)
```

```python
avg_annual_rainfall = annual_climate['annual_rainfall'].mean()

water_results = []

for _, crop in crops.iterrows():

    min_rain = crop['rainfall_opt_min_annual_mm']

    rainfall_gap = max(0, min_rain - avg_annual_rainfall)

    rainfall_coverage = (
        avg_annual_rainfall / min_rain * 100
    )

    water_results.append({
        'crop_name': crop['crop_name'],
        'avg_annual_rainfall': round(avg_annual_rainfall, 2),
        'minimum_rainfall': min_rain,
        'rainfall_gap': round(rainfall_gap, 2),
        'rainfall_coverage_pct': round(rainfall_coverage, 2),
        'irrigation_flag': 'Needed' if rainfall_gap > 0 else 'Review'
    })

water_matching = pd.DataFrame(water_results)

display(water_matching)
```

```python
site_soil_ph = df['soil_ph'].iloc[0]

soil_results = []

for _, crop in crops.iterrows():

    min_ph = crop['soil_ph_opt_min']
    max_ph = crop['soil_ph_opt_max']

    is_suitable = min_ph <= site_soil_ph <= max_ph

    soil_results.append({
        'crop_name': crop['crop_name'],
        'site_soil_ph': round(site_soil_ph, 2),
        'crop_min_ph': min_ph,
        'crop_max_ph': max_ph,
        'soil_ph_match': is_suitable
    })

soil_matching = pd.DataFrame(soil_results)

display(soil_matching)
```

```python
soil_matching['ph_deviation'] = soil_matching.apply(
    lambda row: max(
        row['crop_min_ph'] - row['site_soil_ph'],
        0,
        row['site_soil_ph'] - row['crop_max_ph']
    ),
    axis=1
)

soil_matching = soil_matching.sort_values('ph_deviation')

display(
    soil_matching[
        ['crop_name', 'site_soil_ph',
         'crop_min_ph', 'crop_max_ph', 'ph_deviation']
    ]
)
```

```python
crop_analysis = (
    temperature_matching[
        ['crop_name', 'number_of_months', 'temp_score', 'temp_status']
    ]
    .merge(
        rainfall_matching[
            ['crop_name', 'rainfall_score']
        ],
        on='crop_name'
    )
    .merge(
        water_matching[
            ['crop_name', 'rainfall_gap', 'rainfall_coverage_pct']
        ],
        on='crop_name'
    )
    .merge(
        soil_matching[
            ['crop_name', 'ph_deviation']
        ],
        on='crop_name'
    )
)

crop_analysis['irrigation_needed'] = (
    crop_analysis['rainfall_gap'] > 0
)

display(crop_analysis)
```

```python
water_demand = crops[
    [
        'crop_name',
        'crop_water_need_min_mm_season',
        'crop_water_need_max_mm_season'
    ]
].copy()

water_demand['avg_water_need_mm_season'] = (
    water_demand['crop_water_need_min_mm_season'] +
    water_demand['crop_water_need_max_mm_season']
) / 2

water_demand = water_demand.sort_values(
    'avg_water_need_mm_season'
)

display(water_demand.round(2))
```

```python
crop_analysis = crop_analysis.merge(
    water_demand[
        [
            'crop_name',
            'crop_water_need_min_mm_season',
            'crop_water_need_max_mm_season',
            'avg_water_need_mm_season'
        ]
    ],
    on='crop_name'
)

crop_analysis = crop_analysis.sort_values(
    by='avg_water_need_mm_season'
).reset_index(drop=True)

display(crop_analysis.round(2))
```

```python
def recommend_crops(crop_analysis):

    recommendations = crop_analysis.copy()

    # Rank crops using transparent criteria
    recommendations = recommendations.sort_values(
        by=[
            'temp_score',
            'ph_deviation',
            'avg_water_need_mm_season'
        ],
        ascending=[False, True, True]
    ).reset_index(drop=True)

    recommendations.insert(
        0,
        'rank',
        range(1, len(recommendations) + 1)
    )

    # Add recommendation notes
    def get_reason(row):

        reasons = []

        if row['temp_score'] >= 60:
            reasons.append("Good temperature match")
        elif row['temp_score'] >= 40:
            reasons.append("Moderate temperature match")
        else:
            reasons.append("Limited temperature match")

        if row['ph_deviation'] <= 0.2:
            reasons.append("Close to soil pH range")
        else:
            reasons.append("Soil pH needs review")

        if row['irrigation_needed']:
            reasons.append("Irrigation should be evaluated")

        return " | ".join(reasons)

    recommendations['recommendation_notes'] = (
        recommendations.apply(get_reason, axis=1)
    )

    return recommendations[
        [
            'rank',
            'crop_name',
            'temp_score',
            'ph_deviation',
            'avg_water_need_mm_season',
            'irrigation_needed',
            'recommendation_notes'
        ]
    ]


final_recommendations = recommend_crops(crop_analysis)

display(final_recommendations)
```

```python
output_path = (
    r"G:\FarmWise\Data\Merged_Data\FarmWise_Crop_Recommendations.csv"
)

final_recommendations.to_csv(
    output_path,
    index=False
)

print("Recommendations saved successfully!")
print(output_path)
```

```python
import re
import pandas as pd

# Crop codes and names
crop_names = {
    "cow": "Cowpea",
    "mze": "Maize",
    "pea": "Dry Pea",
    "phb": "Phaseolus Bean",
    "rcd": "Dryland Rice",
    "rcw": "Wetland Rice",
    "whe": "Wheat"
}

gaez_cols = [
    col for col in df.columns
    if col.lower().startswith("gaez_")
    and col.lower().endswith("_suitable_pct")
]

# Extract GAEZ values
gaez_records = []

for col in gaez_cols:
    match = re.match(
        r"gaez_(\w+)_(hilm|hrlm|lilm|lrlm)_suitable_pct",
        col.lower()
    )

    if match:
        crop_code, scenario = match.groups()

        gaez_records.append({
            "crop_code": crop_code,
            "crop_name": crop_names.get(crop_code, crop_code),
            "scenario": scenario.upper(),
            "suitability_pct": df[col].iloc[0]
        })

gaez_df = pd.DataFrame(gaez_records)

display(
    gaez_df.sort_values(["crop_name", "scenario"])
)
```

```python
gaez_summary = (
    gaez_df.groupby(["crop_code", "crop_name"])
    ["suitability_pct"]
    .agg(["mean", "min", "max"])
    .reset_index()
)

gaez_summary = gaez_summary.sort_values(
    "mean",
    ascending=False
).reset_index(drop=True)

gaez_summary.insert(
    0, "gaez_rank", range(1, len(gaez_summary) + 1)
)

display(gaez_summary)
```

```python
# Recalculate GAEZ ranking correctly, preserving ties
gaez_summary["gaez_rank"] = (
    gaez_summary["mean"]
    .rank(method="min", ascending=False)
    .astype(int)
)

# Merge our ranking with GAEZ
comparison = recommendations[
    ["rank", "crop_name", "temp_score",
     "ph_deviation", "avg_water_need_mm_season"]
].merge(
    gaez_summary[["crop_name", "mean", "min", "max", "gaez_rank"]],
    on="crop_name",
    how="left"
)

comparison = comparison.rename(columns={
    "rank": "our_rank",
    "mean": "gaez_mean_pct",
    "min": "gaez_min_pct",
    "max": "gaez_max_pct"
})

comparison["rank_difference"] = (
    comparison["our_rank"] - comparison["gaez_rank"]
)

display(comparison.sort_values("our_rank"))
```

```python
import pandas as pd

# Crop codes and names
crop_mapping = {
    "cow": "Cowpea",
    "mze": "Maize",
    "pea": "Dry Pea",
    "phb": "Phaseolus Bean",
    "rcd": "Dryland Rice",
    "rcw": "Wetland Rice",
    "whe": "Wheat"
}

# Load our recommendation results
recommendations = pd.read_csv(
    r"G:\FarmWise\Data\Merged_Data\FarmWise_Crop_Recommendations.csv"
)

# Calculate average GAEZ suitability for each crop
gaez_results = []

for code, crop_name in crop_mapping.items():

    columns = [
        f"gaez_{code}_hilm_suitable_pct",
        f"gaez_{code}_hrlm_suitable_pct",
        f"gaez_{code}_lilm_suitable_pct",
        f"gaez_{code}_lrlm_suitable_pct"
    ]

    missing = [col for col in columns if col not in df.columns]

    if missing:
        raise ValueError(f"Missing GAEZ columns: {missing}")

    gaez_score = df[columns].iloc[0].mean()

    gaez_results.append({
        "crop_name": crop_name,
        "gaez_average_score": gaez_score
    })

gaez_df = pd.DataFrame(gaez_results)

# Rank GAEZ results (ties receive the same rank)
gaez_df["gaez_rank"] = (
    gaez_df["gaez_average_score"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# Compare both ranking systems
comparison = recommendations[
    ["crop_name", "rank", "overall_score"]
].merge(
    gaez_df,
    on="crop_name"
)

comparison = comparison.rename(columns={
    "rank": "farmwise_rank",
    "overall_score": "farmwise_score"
})

comparison["rank_difference"] = (
    comparison["farmwise_rank"] - comparison["gaez_rank"]
)

comparison = comparison.sort_values("farmwise_rank")

display(comparison.round(2))
```

```python
# Calculate GAEZ ranks with average ranking for ties
comparison["gaez_rank_average"] = (
    comparison["gaez_average_score"]
    .rank(ascending=False, method="average")
)

# Rank difference using average ranks
comparison["rank_difference_average"] = (
    comparison["farmwise_rank"]
    - comparison["gaez_rank_average"]
)

# Mean Absolute Rank Difference
mean_rank_difference = (
    comparison["rank_difference_average"].abs().mean()
)

# Spearman rank correlation
spearman = comparison["farmwise_score"].corr(
    comparison["gaez_average_score"],
    method="spearman"
)

print("Validation Metrics")
print("=" * 40)
print(f"Mean Absolute Rank Difference: {mean_rank_difference:.3f}")
print(f"Spearman Rank Correlation: {spearman:.3f}")

display(
    comparison[
        ["crop_name", "farmwise_rank",
         "gaez_rank_average", "rank_difference_average"]
    ].round(2)
)

# Save validation results
comparison.to_csv(
    r"G:\FarmWise\Data\Merged_Data\FarmWise_GAEZ_Validation.csv",
    index=False
)

print("\nValidation results saved successfully!")
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

# Import project functions
from recommendation_service import load_data
from crop_suitability import analyze_crop_suitability
from scoring import rank_crops

# 1) Load project data
unified_path = r"G:\FarmWise\Data\Merged_Data\FarmWise_Unified_Dataset.csv"
crops_path = r"G:\FarmWise\Data\Crop_Requirements\FarmWise_Crop_Requirements_Complete.csv"

df, crops = load_data(unified_path, crops_path)

# 2) Run recommendation pipeline
suitability = analyze_crop_suitability(df, crops)
recommendations = rank_crops(suitability)

# 3) Validate input data
assert not df.empty, "Unified dataset is empty"
assert not crops.empty, "Crop requirements dataset is empty"
assert len(crops) == 7, f"Expected 7 crops, got {len(crops)}"

# 4) Validate recommendation output
assert len(recommendations) == len(crops), "Some crops are missing"
assert "overall_score" in recommendations.columns
assert "rank" in recommendations.columns

crop_col = next(
    (col for col in ["crop_name", "crop", "Crop"]
     if col in recommendations.columns),
    None
)

assert crop_col is not None, "Crop name column was not found"
assert recommendations[crop_col].is_unique, "Duplicate crops"
assert recommendations["overall_score"].notna().all()
assert np.isfinite(recommendations["overall_score"]).all()
assert recommendations["overall_score"].between(0, 100).all()

# 5) Validate ranking order
recommendations = recommendations.sort_values(
    "rank"
).reset_index(drop=True)

assert recommendations["rank"].tolist() == list(
    range(1, len(crops) + 1)
)

assert recommendations["overall_score"].is_monotonic_decreasing

# 6) Validate explainability fields
explanation_columns = [
    "strengths",
    "limitations",
    "recommendation_notes",
    "recommendation_summary"
]

for col in explanation_columns:
    assert col in recommendations.columns, f"Missing column: {col}"
    assert recommendations[col].notna().all(), f"Empty values in {col}"

# 7) Save recommendations as CSV
output_path = Path(
    r"G:\FarmWise\Data\Merged_Data\FarmWise_Crop_Recommendations.csv"
)

output_path.parent.mkdir(parents=True, exist_ok=True)

recommendations.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

# 8) Verify saved CSV
saved_recommendations = pd.read_csv(output_path)

assert len(saved_recommendations) == len(recommendations)
assert saved_recommendations[crop_col].is_unique

print("ALL FINAL TESTS PASSED")
print(f"CSV successfully saved: {output_path}")
print(f"Number of recommendations: {len(saved_recommendations)}")

display(saved_recommendations)
```
