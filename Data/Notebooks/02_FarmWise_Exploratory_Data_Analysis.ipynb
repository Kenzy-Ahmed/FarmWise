```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

df = pd.read_csv("/content/FarmWise_Unified_Dataset.csv")

df["date"] = pd.to_datetime(df["date"])

print("Dataset Shape:", df.shape)
print("Date Range:", df["date"].min(), "to", df["date"].max())

display(df.head())
```

```python
# 1. Missing values analysis

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

missing = missing.sort_values("Missing_Percentage", ascending=False)

print("MISSING VALUES ANALYSIS")
display(missing[missing["Missing_Count"] > 0].round(2))


# 2. Duplicate rows

print("\nDuplicate Rows:", df.duplicated().sum())


# 3. Constant columns

constant_columns = [
    col for col in df.columns
    if df[col].nunique(dropna=True) <= 1
]

print("\nConstant Columns:")
print(constant_columns)


# 4. Unique values in important groups

print("\nUNIQUE VALUES")

print("Unique dates:", df["date"].nunique())
print("Unique NDVI composite dates:", df["ndvi_composite_date"].nunique())
print("Unique soil profiles:", df.filter(regex="^soil_").nunique().to_dict())
print("Unique GAEZ values:", df.filter(regex="^gaez_").nunique().to_dict())
```

```python
# Prepare one row per NDVI composite

ndvi_df = (
    df.dropna(subset=["ndvi_composite_date", "ndvi_mean"])
    [["ndvi_composite_date", "ndvi_mean", "ndvi_median",
      "ndvi_std", "ndvi_low_coverage_flag"]]
    .drop_duplicates(subset="ndvi_composite_date")
    .sort_values("ndvi_composite_date")
    .copy()
)

ndvi_df["ndvi_composite_date"] = pd.to_datetime(
    ndvi_df["ndvi_composite_date"]
)

print("Unique NDVI Composites:", len(ndvi_df))
print("Low Coverage Composites:",
      ndvi_df["ndvi_low_coverage_flag"].sum())

# 1. NDVI Distribution

plt.figure(figsize=(10, 5))

plt.hist(ndvi_df["ndvi_mean"], bins=20, edgecolor="black")

plt.title("Distribution of NDVI Mean Values")
plt.xlabel("NDVI Mean")
plt.ylabel("Number of Composites")

plt.show()


# 2. NDVI Trend Over Time

plt.figure(figsize=(14, 5))

plt.plot(
    ndvi_df["ndvi_composite_date"],
    ndvi_df["ndvi_mean"],
    marker="o",
    markersize=3,
    linewidth=1
)

plt.title("NDVI Trend Over Time (2015–2025)")
plt.xlabel("Date")
plt.ylabel("NDVI Mean")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 3. Annual Average NDVI

ndvi_df["year"] = ndvi_df["ndvi_composite_date"].dt.year

annual_ndvi = (
    ndvi_df.groupby("year")["ndvi_mean"]
    .mean()
    .reset_index()
)

display(annual_ndvi.round(4))

plt.figure(figsize=(12, 5))

plt.plot(
    annual_ndvi["year"],
    annual_ndvi["ndvi_mean"],
    marker="o"
)

plt.title("Annual Average NDVI (2015–2025)")
plt.xlabel("Year")
plt.ylabel("Average NDVI")

plt.grid(True)
plt.show()
```

```python
# Monthly NDVI Analysis

ndvi_df["month"] = ndvi_df["ndvi_composite_date"].dt.month

monthly_ndvi = (
    ndvi_df.groupby("month")["ndvi_mean"]
    .agg(["mean", "median", "std", "count"])
    .reset_index()
)

display(monthly_ndvi.round(4))

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_ndvi["month"],
    monthly_ndvi["mean"],
    marker="o"
)

plt.title("Average Monthly NDVI")
plt.xlabel("Month")
plt.ylabel("Average NDVI")

plt.xticks(range(1, 13))
plt.grid(True)
plt.show()


# Seasonal NDVI Distribution

plt.figure(figsize=(12, 5))

sns.boxplot(
    data=ndvi_df,
    x="month",
    y="ndvi_mean"
)

plt.title("Monthly NDVI Distribution")
plt.xlabel("Month")
plt.ylabel("NDVI Mean")

plt.show()
```

```python
# Prepare weather data

weather_columns = [
    "power_t2m_mean_c",
    "power_precip_mm_day",
    "power_rh2m_pct",
    "power_wind2m_ms",
    "gpm_precip_mm_day",
    "smap_soil_moisture_am_m3m3"
]

weather_df = df.copy()

weather_df["date"] = pd.to_datetime(weather_df["date"])

weather_rows = []

for _, row in ndvi_df.iterrows():

    target_date = row["ndvi_composite_date"]

    previous_period = weather_df[
        (weather_df["date"] >= target_date - pd.Timedelta(days=16)) &
        (weather_df["date"] < target_date)
    ]

    result = {
        "composite_date": target_date,
        "ndvi_mean": row["ndvi_mean"]
    }

    for col in weather_columns:
        result[col] = previous_period[col].mean()

    weather_rows.append(result)

weather_analysis = pd.DataFrame(weather_rows)

display(weather_analysis.head())

print("Shape:", weather_analysis.shape)
```

```python
correlation_data = weather_analysis.drop(
    columns=["composite_date"]
)

correlation_matrix = correlation_data.corr(method="pearson")

ndvi_correlation = (
    correlation_matrix["ndvi_mean"]
    .drop("ndvi_mean")
    .sort_values(key=abs, ascending=False)
)

print("Correlation with NDVI:")
display(ndvi_correlation.to_frame("Correlation").round(3))

plt.figure(figsize=(11, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Weather and Soil Moisture Correlation with NDVI")
plt.tight_layout()
plt.show()
```

```python
correlation_check = []

for col in weather_columns:
    temp = weather_analysis[["ndvi_mean", col]].dropna()

    correlation_check.append({
        "Feature": col,
        "Valid_Observations": len(temp),
        "Pearson": temp["ndvi_mean"].corr(temp[col], method="pearson"),
        "Spearman": temp["ndvi_mean"].corr(temp[col], method="spearman")
    })

correlation_check_df = pd.DataFrame(correlation_check)

display(
    correlation_check_df
    .sort_values("Pearson", key=abs, ascending=False)
    .round(3)
)
```

```python
outlier_columns = [
    "ndvi_mean",
    "power_t2m_mean_c",
    "power_precip_mm_day",
    "power_rh2m_pct",
    "power_wind2m_ms",
    "gpm_precip_mm_day",
    "smap_soil_moisture_am_m3m3"
]

outlier_results = []

for col in outlier_columns:
    values = weather_analysis[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = values[(values < lower) | (values > upper)]

    outlier_results.append({
        "Feature": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": len(outliers) / len(values) * 100
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df.round(3))
```

```python
plt.figure(figsize=(14, 8))

for i, col in enumerate(outlier_columns, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=weather_analysis[col])
    plt.title(col, fontsize=9)

plt.tight_layout()
plt.show()
```

```python
for col in [
    "power_precip_mm_day",
    "gpm_precip_mm_day",
    "smap_soil_moisture_am_m3m3"
]:
    values = weather_analysis[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = weather_analysis[
        (weather_analysis[col] < lower) |
        (weather_analysis[col] > upper)
    ][["composite_date", col, "ndvi_mean"]]

    print(f"\n{col} — Outlier Values")
    display(outliers.sort_values(col, ascending=False).head(15))
```

```python
rain_comparison = weather_analysis[
    ["composite_date", "power_precip_mm_day", "gpm_precip_mm_day"]
].dropna().copy()

rain_comparison["Difference"] = (
    rain_comparison["power_precip_mm_day"] -
    rain_comparison["gpm_precip_mm_day"]
).abs()

print("Rainfall Correlation:")
print(
    rain_comparison["power_precip_mm_day"].corr(
        rain_comparison["gpm_precip_mm_day"]
    )
)

display(
    rain_comparison.sort_values(
        "Difference", ascending=False
    ).head(10).round(3)
)
```

# FarmWise Feature Selection Report

## 1. Dataset Overview

- Dataset: FarmWise_Unified_Dataset.csv
- Records: 4018
- Features: 61 columns
- Study Area: El Zankalon, Sharqia, Egypt
- Time Range: 2015–2025

## 2. Feature Groups

| Group | Number of Features | Role |
|---|---:|---|
| Weather | 6 | Environmental conditions |
| Rainfall | 2 | Water availability |
| Soil | 8 | Soil properties |
| SMAP | 2 | Soil moisture information |
| GAEZ | 28 | Crop suitability reference |
| NDVI | 9 | Vegetation monitoring |

## 3. Feature Selection Decisions

### Weather
Retained as dynamic environmental features.

### Rainfall
POWER and GPM retained as separate sources.
Missing GPM values should be documented.

### Soil
All 8 features retained as static soil information.
They have constant values within the current study area.

### SMAP
Retained as supplementary information.
Both columns have 66.45% missing values.

### GAEZ
All 28 features retained as crop suitability references.
They are static within the current study area.

### NDVI
Retained for vegetation monitoring.
ndvi_mean may be used for NDVI prediction, but not as a direct crop recommendation label.

## 4. Important Limitations

- Soil and GAEZ features do not represent spatial variation within the study area.
- SMAP has a high missing-value percentage.
- NDVI observations are composite-based, not independent daily measurements.
- No verified crop-label target is currently available.

## 5. Final Decision

The selected features will support the crop recommendation
workflow through environmental information and crop suitability references.

The final recommendation input schema should be coordinated
with the Crop Recommendation team.
